```python
from __future__ import annotations
import random
import time
from comp271_mystery_sorters import gandalf, saruman, sauron


def generate_random_array(size: int, min_value: int = 0, max_value: int = 100) -> list[int]:
    """Generate a random array of integers.

    Args:
        size (int): The size of the array to generate.
        min_value (int): The minimum value for the random integers.
        max_value (int): The maximum value for the random integers.

    Returns:
        list[int]: A list containing random integers.
    """
    return [random.randint(min_value, max_value) for _ in range(size)]

def time_function(func, *args, **kwargs) -> float:
    """Measure the execution time of a function.

    Args:
        func (callable): The function to measure.
        *args: Positional arguments to pass to the function.
        **kwargs: Keyword arguments to pass to the function.

    Returns:
        float: The time taken to execute the function, in seconds.
    """
    start_time = time.perf_counter()
    func(*args, **kwargs)
    end_time = time.perf_counter()
    return end_time - start_time

if __name__ == "__main__":
    functions_to_test = [gandalf, saruman, sauron]
    max_size = 256
    step = 2
    # Safety valve: if one function's runtime blows up as size grows,
    # stop increasing its size instead of letting the script stall.
    max_seconds_per_call = 5.0

    for function in functions_to_test:
        for size in range(1, max_size + 1, step):
            random_array = generate_random_array(size)
            time_taken = time_function(function, random_array)
            print(f"Function: {function.__name__}, Size: {size}, Time taken: {time_taken:.6f} s")
            if time_taken > max_seconds_per_call:
                print(f"  -> {function.__name__} exceeded {max_seconds_per_call}s budget; skipping larger sizes")
                break  #  use of break is highly discouranged, 
                       # but here it is used to prevent the script 
                       # from stalling on large inputs
```
