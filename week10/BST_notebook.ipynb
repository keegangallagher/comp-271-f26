```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

    def __str__(self):
        return str(self.value)

```

```python
h = Node("Hello")
w = Node("World")
e = Node("Hello")

print(h == w)  
print(h == e)  
print(h > w)   
print(h < w)   
```
