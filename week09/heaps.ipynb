# Heaps (Priority Queues)

A priority queue is a data structure with two, perhaps three, user-facing functions:
* **get:** always removes and returns the smallest (or the largest) value in the collection.
* **add:** places a new value in the collection
* **peek:** previews what's the smallest (or the largest) value in the collection.

For this note, we'll focus on priority queues that return the smallest value.

A *trivial* way to implement a priority queue is to store the data in an ``array`` and keep the array sorted in ascending order. This guarrantees that `array[0]` is the smallest value. As we apply the `get` method, array elements must be shifted to the left so that we always remove from position `[0]`; this would require $\Theta (n)$ time. 

Alternatively, we can have a integer pointer to the next element to remove from the sorted array, moving from `[0]`, to `[1]`, etc. This is a $\mathcal O(1)$ operation. However sorting the ``array`` takes $\Theta(n\log n)$ time every time we add an element to it.

If we do not sort the ``array`` but instead we perform a linear scan to find its minimum value, we can obtain it in $\Theta (n)$ which is faster than sorting.


There is even a faster way to obtain the same results. If the values are stored as a **complete binary tree,** data can be reordered in $\Theta (\log n)$ ensuring that the smallest value is always at the root of the tree. 

> **Definition:** A binary tree in which every level (depth), except possibly the deepest, is completely filled. At the last level, all nodes must be as far left as possible.</br>
> Paul E. Black, "[complete binary tree](https://xlinux.nist.gov/dads/HTML/completeBinaryTree.html)", in Dictionary of Algorithms and Data Structures, 2016

A complete binary tree that guarantees its root is the smallest value is called a *minimum heap.* When the root is largest value, it's called *maximum heap.*

We'll focus on the minimum heap for now.

A minimum heap is a complete binary tree where 

* **the value of each parent node is less than or equal to the values of both its children.**  This is called the *minimum heap property;*
* a new value is added as the last leaf node in the tree and if the minimum heap property is violated, it must be restored;
* the root is always the smalest value in the tree and the value to be removed and returned during a `get` operation. The last leaf of the heap is moved to the vacant root. If the minimum heap property is violated, it must be restored;
* a violation of the heap property is restored by successive swaps between pairs of offending nodes.

## Adding data to a minimum heap

Adding data to a minimum heap is shown below. 

* **(i):** Initial state of the heap with 4 elements. Each node in this heap complies with the minimum heap property.
* **(ii):** Suppose we want to add the value $0$ to the heap. Initially it is added as the last leaf node of the complete binary tree. Immediately we notice that this new node and its parent node violate the minimum haap property. 
* **(iii):** We begin restoring the minimum heap property by swapping the newly added node and its parent. The new node moves up a level in the tree, but the minimum heap property is still violated.
* **(iv):** We continue restoring the minimum heap property by swapping the offeding notes. The property is finally restored.

In the worst case scenario, a new node at the bottom of the tree will have to climb all the way to the top, to restore the minimum heap property. For a complete binary tree, this requires $\Theta (\log n)$ steps because that's the height of the tree.

## Removing data from a minimum heap

Data from a minimum heap are always removed from the root -- that's where the data structure places its smallest element. The figure below shows what happens after the removal of the smallest element.

* **(i):** Initial state of the heap with 5 elements. Each node in this heap complies with the minimum heap property.
* **(ii):** The smallest element (always the root) has been removed and the root is now vacant. We replace the content of the root with the value of the last node of the complete binary tree.
* **(iii):** Moving that value to the root may violate the minimum heap property. In this case we swap the offending node with the smallest of its two children.
* **(iv):** After this swap, there are no more violations of the minimum heap property.

In the worst case scenario, the last value of the tree that we moved to the root node may have to climb all the way to the bottom, to restore the minimum heap property. For a complete binary tree, this requires $\Theta (\log n)$ steps because that's the height of the tree.

## Moving data around a tree

A complete binary tree with depth $L$ has anywhere between $2^L-1$ and $2^{L-1}$ nodes. Therefore we can write that $L\in\Theta(\log_2 n)$ and for practical purposes:

```python
def L(n:int) -> int:
    """Given the number of nodes in a complete binary tree, return the depth of the tree using log base 2."""
    if n <= 0:
        raise ValueError("Number of nodes must be a positive integer.")
    import math
    return math.floor(math.log2(n)) + 1
```

Going up and down a binary tree may be fast but also requires significant memory to store the node objects and the left/right pointers. Essentially we are trading off time complexity for space complexity. There is a way to achieve the speed efficiency of the complete binary tree without sacrificing space (memory). It involves a humble array. This is shown below.

![](https://raw.githubusercontent.com/lgreco/images/refs/heads/main/graphs/Complete%20Binary%20Tree.drawio.png)


The tree is traversed level-by-level, from left to right, and its values are added to an array. For a tree with depth $L$ we know that there will be at most $2^{L-1}$ values so, we can size the array accordingly. The parent-children relationships can be preserved with the simple arithmetic below, where $n$ is the actual number of nodes in the tree and therefore the size of the array.

\begin{align}
\text{left}_\text{child}(\text{parent}) & = 2\times\text{parent}+1, && \ \text{parent} < n-1 \\
\text{right}_\text{child}(\text{parent}) & = \text{left}_\text{child}(\text{parent}) +1, && \ \text{parent} < n-2 \\
\text{parent}(\text{child}) & = \left\lfloor \frac{\text{child}-1}{2} \right\rfloor, && \ \text{child} > 0
\end{align}

The floor operator $\lfloor\,\rfloor$ is equivalent to *float-to-integer truncation* that is available in most programming languages. In Python, the truncation is implemented with the function `int()` but for *positive numbers* we can also obtain the same result with integer division (`//`). The simple arithmetic above is straight forward to code.

```python


def left_child(self, parent: int) -> int:
    """Return the index of the left child of the given parent index."""
    return 2 * parent + 1


def right_child(self, parent: int) -> int:
    """Return the index of the right child of the given parent index."""
    return left_child(parent) + 1


def parent(self, child: int) -> int:
    """Return the index of the parent of the given child index."""
    return (child - 1) // 2
```
