# Heaps (Priority Queues)

A priority queue is a data structure with two, perhaps three, user-facing functions:
* **get:** always removes and returns the smallest (or the largest) value in the collection.
* **add:** places a new value in the collection
* **peek:** previews what's the smallest (or the largest) value in the collection.

For this note, we'll focus on priority queues that return the smallest value.

A *trivial* way to implement a priority queue is to store the data in an ``array`` and keep the array sorted in ascending order. This guarrantees that `array[0]` is the smallest value. As we apply the `get` method, array elements must be shifted to the left so that we always remove from position `[0]`; this would require $\Theta (n)$ time. 

Alternatively, we can have a integer pointer to the next element to remove from the sorted array, moving from `[0]`, to `[1]`, etc. This is a $\mathcal O(1)$ operation. However sorting the ``array`` takes $\Theta(n\log n)$ time every time we add an element to it.

If we do not sort the ``array`` but instead we perform a linear scan to find its minimum value, we can obtain it in $\Theta (n)$ which is faster than sorting.


There is even a faster way to obtain the same results. If the values are stored as a complete binary tree, data can be reordered in $\Theta (\log n)$ ensuring that the smallest value is always at the root of the tree.

A complete binary tree that guarantees its root is the smallest value is called a *minimum heap.* When the root is largest value, it's called *maximum heap.*

We'll focus on the minimum heap for now.

A minimum heap is a complete binary tree where 

* **the value of each parent node is less than or equal to the values of both its children.**  This is called the *minimum heap property;*
* a new value is added as the last leaf node in the tree and if the minimum heap property is violated, it must be restored;
* the root is always the smalest value in the tree and the value to be removed and returned during a `get` operation. The last leaf of the heap is moved to the vacant root. If the minimum heap property is violated, it must be restored;
* a violation of the heap property is restored by successive swaps between pairs of offending nodes.

## Adding data to a minimum heap

Adding data to a minimum heap is shown below. 

* **(i):** Initial state of the heap with 4 elements. Each node in this heap complies with the minimum heap property.
* **(ii):** Suppose we want to add the value $0$ to the heap. Initially it is added as the last leaf node of the complete binary tree. Immediately we notice that this new node and its parent node violate the minimum haap property. 
* **(iii):** We begin restoring the minimum heap property by swapping the newly added node and its parent. The new node moves up a level in the tree, but the minimum heap property is still violated.
* **(iv):** We continue restoring the minimum heap property by swapping the offeding notes. The property is finally restored.

In the worst case scenario a new node at the bottom of the tree will have to climb all the way to the top, to restore the minimum heap property. For a complete binary tree, this will require $\Theta (\log n)$ steps because that's the height of the tree.

```python


def left_child(self, parent: int) -> int:
    """Return the index of the left child of the given parent index."""
    return 2 * parent + 1


def right_child(self, parent: int) -> int:
    """Return the index of the right child of the given parent index."""
    return left_child(parent) + 1


def parent(self, child: int) -> int:
    """Return the index of the parent of the given child index."""
    return (child - 1) // 2


def swap(array: list, i: int, j: int) -> None:
    """Swap the elements at indices i and j in the given array."""
    temp = array[i]
    array[i] = array[j]
    array[j] = temp


def heapify_down(array: list, parent: int) -> None:
    """Heapify down the element at the given index in the array."""
    n = len(array)
    while left_child(parent) < n:
        left = left_child(parent)
        right = right_child(parent)
        smallest = left
        if right < n and array[right] < array[left]:
            smallest = right
        if array[parent] > array[smallest]:
            swap(array, parent, smallest)
            parent = smallest
```
