## Part 1 — `MinHeap271`

### The interface

A priority queue has two basic operations and two auxiliary ones. All four
work under one invariant: **`get` always returns the smallest value in the
collection.**

| Method | Success | Failure (nothing changes) |
|---|---|---|
| `add(value)` | places `value` in the heap | — (`add` cannot fail) |
| `get()` | removes and returns the smallest value | heap is empty → returns `None` |
| `peek()` | returns the smallest value **without** removing it | heap is empty → returns `None` |
| `size()` | returns the number of values in the heap | — |

### The representation

The heap is a complete binary tree stored in a Python list, `self.storage`,
level by level, left to right. The root is at `storage[0]` and the last leaf
is at `storage[len(storage) - 1]`. Parents and children are found with the
arithmetic from class:

\begin{align}
\text{left}(p) & = 2p+1 \\
\text{right}(p) & = 2p+2 \\
\text{parent}(c) & = \left\lfloor \frac{c-1}{2} \right\rfloor
\end{align}

These three are given, along with `_swap`. Unlike last week's queue and
stack, this storage **does** grow and shrink: `add` appends to the end of
the list and `get` removes the last element. Those are the only two places
the list changes size.

### What you write

- **`_is_leaf(position)`.** In class I suggested an auxiliary function that
  tells whether a given array position is a leaf. A position is a leaf when
  it has no children in the array. Which of its two children is enough to
  check?
- **`_bubble_up()`.** Called by `add`, right after the new value has been
  appended. Track the new value's position, starting at `len - 1`. While it
  is not at the root and it is smaller than its parent, swap it with its
  parent and follow it up.
- **`_sink_down()`.** Called by `get`, right after the last value has been
  moved into the vacant root. Track that value's position, starting at `0`.
  While it is not at a leaf, find its **smaller** child: assume the left
  child, then switch to the right child if the right child exists and is
  smaller. If the value is greater than that smaller child, swap them and
  follow the value down. Otherwise, stop.
- **`add`, `get`, `peek`, `size`.** `add` and `get` are short: put the value
  in the right place, then call the right restoring operation.

Before you code `get`, trace it by hand on a heap that holds **exactly one**
value. After you remove the last element of the list, what is there to move
into the root?

### Rules

- No `import heapq`, and no `sorted()`, `.sort()`, `min()` or `max()` over
  `storage`. The whole point is that the heap finds the minimum by itself.
- Compare values only with `<` and `>`. That way the heap works for any

### Tests for Part 1

`test_heap_property_holds` checks the array itself after every operation:
every parent must be less than or equal to both its children. If your `get`
returns the right values but this test fails, your heap is getting the right
answers by accident.

````python
# Given -- tests for Part 1. Run this cell after every change.
import inspect
import random


def heap_property_ok(heap: MinHeap271) -> bool:
    """True if every parent in heap.storage is <= both of its children."""
    a = heap.storage
    ok = True
    for child in range(1, len(a)):
        if a[(child - 1) // 2] > a[child]:
            ok = False
    return ok


def test_heap_empty():
    h = MinHeap271()
    assert h.size() == 0, "a new heap has size 0"
    assert h.peek() is None, "peek on an empty heap returns None"
    assert h.get() is None, "get on an empty heap returns None"
    assert h.size() == 0, "a failed get changes nothing"


def test_heap_is_leaf():
    h = MinHeap271()
    h.storage = [1, 2, 3, 4]          # set directly: tests _is_leaf alone
    assert not h._is_leaf(0), "the root has two children"
    assert not h._is_leaf(1), "position 1 has a left child at 3"
    assert h._is_leaf(2), "position 2 would have children at 5 and 6"
    assert h._is_leaf(3), "position 3 is the last leaf"


def test_heap_single_value():
    h = MinHeap271()
    h.add(42)
    assert h.size() == 1
    assert h.peek() == 42, "peek returns the only value"
    assert h.size() == 1, "peek does not remove"
    assert h.get() == 42, "get returns the only value"
    assert h.size() == 0, "and the heap is empty again"
    assert h.get() is None


def test_heap_bubble_up_class_example():
    # From class: 2, 8, 4 is a minimum heap; adding 1 floats to the root.
    h = MinHeap271()
    for v in [2, 8, 4]:
        h.add(v)
    h.add(1)
    assert h.storage == [1, 2, 4, 8], f"expected [1, 2, 4, 8], got {h.storage}"


def test_heap_sink_down_swaps_with_smaller_child():
    # 3 moves to the root and must swap with 1 (the smaller child), not 2.
    h = MinHeap271()
    h.storage = [0, 2, 1, 3]          # a valid minimum heap
    assert h.get() == 0
    assert h.storage == [1, 2, 3], f"expected [1, 2, 3], got {h.storage}"


def test_heap_ascending_and_descending_adds():
    for values in ([1, 2, 3, 4, 5, 6, 7], [7, 6, 5, 4, 3, 2, 1]):
        h = MinHeap271()
        for v in values:
            h.add(v)
        out = [h.get() for _ in values]
        assert out == [1, 2, 3, 4, 5, 6, 7], f"added {values}, got {out}"


def test_heap_duplicates():
    h = MinHeap271()
    for v in [2, 1, 2, 1, 2]:
        h.add(v)
    out = [h.get() for _ in range(5)]
    assert out == [1, 1, 2, 2, 2], f"got {out}"


def test_heap_property_holds():
    random.seed(271)
    h = MinHeap271()
    for _ in range(300):
        if random.random() < 0.6:
            h.add(random.randint(-50, 50))
        else:
            h.get()
        assert heap_property_ok(h), f"heap property broken: {h.storage}"


def test_heap_random_matches_sorted():
    random.seed(9)
    for n in [2, 3, 10, 31, 32, 33, 200]:
        values = [random.randint(0, 100) for _ in range(n)]
        h = MinHeap271()
        for v in values:
            h.add(v)
        assert h.size() == n
        out = [h.get() for _ in range(n)]
        assert out == sorted(values), f"n={n}: values came out out of order"
        assert h.get() is None


def test_heap_tuples():
    # Part 2 stores (tier, arrival_number, ...) tuples. Python compares tuples
    # element by element, so a smaller tier wins and ties go to the earlier arrival.
    h = MinHeap271()
    for patient in [(3, 1), (2, 2), (1, 3), (2, 4), (1, 5)]:
        h.add(patient)
    out = [h.get() for _ in range(5)]
    assert out == [(1, 3), (1, 5), (2, 2), (2, 4), (3, 1)], f"got {out}"


def test_heap_no_shortcuts():
    # Read each method's source. (inspect.getsource on a class defined in a
    # notebook can fail; on its functions it works.)
    src = "".join(inspect.getsource(member) for member in vars(MinHeap271).values()
                  if inspect.isfunction(member))
    for banned in ["heapq", "sorted(", ".sort(", "min(", "max("]:
        assert banned not in src, f"MinHeap271 may not use {banned}"


run_tests("test_heap")
````

## Part 2 — An emergency room

### The model

An emergency room is a priority queue with people in it. Patients do not
get seen in the order they arrive; they get seen in the order of how badly
they need to be seen. We classify every patient into one of three tiers. A
smaller number means more urgent, which is exactly what a **minimum** heap
hands out first:

| Tier | Meaning | Who comes out of the heap first |
|---|---|---|
| `1` | life-threatening | always ahead of tiers 2 and 3 |
| `2` | severe | ahead of tier 3 |
| `3` | non-life-threatening | last. You may wait 12 hours with a fever. |

Within a tier, patients are seen first come, first served. We get that for
free by storing each waiting patient as a tuple:

```python
(tier, arrival_number, arrival_minute)
```

Python compares tuples element by element. Two patients compare by `tier`
first, and only when the tiers are equal does `arrival_number` decide. Every
patient has a different `arrival_number` (0, 1, 2, … in order of arrival),
so the comparison never needs to look at `arrival_minute`.
`test_heap_tuples` above already checked that your heap handles this.

### The clock

A discrete event simulation follows a system through the moments when
something happens: a patient arrives, a doctor finishes a treatment, the
next patient is called in. The simplest way to run one is to advance a
clock **one minute at a time**. Each minute, in this order:

1. **Arrivals.** With probability `arrival_prob`, one new patient walks in.
   Give them a random tier and the next `arrival_number`, and `add` them to
   the waiting room (your `MinHeap271`).
2. **Treatment.** Every doctor who is free this minute calls in the next
   patient with `get`, if anyone is waiting. That patient waited
   `minute - arrival_minute` minutes. Record it under their tier. The doctor
   is now busy for a random number of minutes that depends on the tier.

A doctor who starts a treatment at minute `m` that lasts `t` minutes is free
again at minute `m + t`. One list, with one entry per doctor holding *the
minute this doctor is next free*, is all the bookkeeping the doctors need.

When the clock runs out, some patients may still be in the waiting room.
They were never seen, so they have no wait time to average. **Count them
anyway.** A tier with a short average wait and 400 people still in the
waiting room is not a tier that is being served well.

### The parameters

We will research realistic numbers together: how many ER visits are
life-threatening on a given day, how many are severe, and how long each kind
of treatment takes. Until then, the values below are **placeholders** that
behave sensibly. They are biased toward the non-life-threatening and severe
tiers, as real ERs are. Your code must read them from these names; never
type the numbers into your functions. Then the simulation keeps working when
we replace them.

````python
# Given -- simulation settings. PLACEHOLDER values until we research real ones.
LIFE_THREATENING = 1
SEVERE = 2
NON_LIFE_THREATENING = 3
TIERS = [LIFE_THREATENING, SEVERE, NON_LIFE_THREATENING]

# Share of arrivals in each tier, in the same order as TIERS. Must add up to 1.
TIER_WEIGHTS = [0.10, 0.35, 0.55]

# Treatment length in minutes for each tier: (shortest, longest), inclusive.
TREATMENT_MINUTES = {
    LIFE_THREATENING: (60, 180),
    SEVERE: (30, 90),
    NON_LIFE_THREATENING: (10, 30),
}

ARRIVAL_PROB = 0.25          # chance of one arrival in any given minute (about 15 per hour)
NUM_DOCTORS = 12
SIM_MINUTES = 7 * 24 * 60    # one week
SEED = 271                   # same seed -> same "random" week, so runs can be compared
````

### What you write

Three small functions, then the simulation itself. The three small ones are
where all the randomness lives. `simulate` should not call anything from
`random` except `random.seed`.

`simulate` returns its results as a dictionary with one entry per tier:

```python
{
    LIFE_THREATENING:     {"arrived": ..., "seen": ..., "total_wait": ..., "still_waiting": ...},
    SEVERE:               {...},
    NON_LIFE_THREATENING: {...},
}
```

- `arrived`: how many patients of this tier walked in.
- `seen`: how many were called in by a doctor.
- `total_wait`: the sum of the waits of the patients who were seen.
- `still_waiting`: how many were still in the waiting room when the clock ran out.

For every tier, `arrived == seen + still_waiting`. Nobody leaves the ER
without being counted.

````python
# Given -- turning statistics into a report.
TIER_NAMES = {LIFE_THREATENING: "life-threatening",
              SEVERE: "severe",
              NON_LIFE_THREATENING: "non-life-threatening"}


def average_wait(stats: dict, tier: int) -> float | None:
    """Average wait, in minutes, of the patients of `tier` who were seen."""
    average = None
    if stats[tier]["seen"] > 0:
        average = stats[tier]["total_wait"] / stats[tier]["seen"]
    return average


def print_report(stats: dict) -> None:
    print(f"{'tier':22s} {'arrived':>8s} {'seen':>6s} {'waiting':>8s} {'avg wait (min)':>15s}")
    for tier in TIERS:
        s = stats[tier]
        avg = average_wait(stats, tier)
        shown = "--" if avg is None else f"{avg:.1f}"
        print(f"{TIER_NAMES[tier]:22s} {s['arrived']:8d} {s['seen']:6d} "
              f"{s['still_waiting']:8d} {shown:>15s}")


print_report(simulate(NUM_DOCTORS, ARRIVAL_PROB, SIM_MINUTES, SEED))
````

### Tests for Part 2

Randomness makes exact answers impossible to test, so these tests check
things that must be true no matter what the dice say: counts add up,
arrivals stop when `arrival_prob` is 0, the most urgent patients wait less
than the least urgent ones when doctors are scarce, and the same seed gives
the same week.

````python
# Given -- tests for Part 2. Run this cell after every change.
def test_er_random_tier_follows_weights():
    random.seed(1)
    picks = [random_tier() for _ in range(20_000)]
    for tier, weight in zip(TIERS, TIER_WEIGHTS):
        share = picks.count(tier) / len(picks)
        assert abs(share - weight) < 0.02, f"tier {tier}: share {share:.3f}, expected about {weight}"


def test_er_arrival_happens_extremes_and_rate():
    random.seed(2)
    assert not any(arrival_happens(0.0) for _ in range(1000)), "probability 0 never happens"
    assert all(arrival_happens(1.0) for _ in range(1000)), "probability 1 always happens"
    share = sum(arrival_happens(0.25) for _ in range(20_000)) / 20_000
    assert abs(share - 0.25) < 0.02, f"probability 0.25 happened {share:.3f} of the time"


def test_er_treatment_minutes_in_range():
    random.seed(3)
    for tier in TIERS:
        low, high = TREATMENT_MINUTES[tier]
        seen = {treatment_minutes(tier) for _ in range(5000)}
        assert min(seen) >= low and max(seen) <= high, f"tier {tier}: out of [{low}, {high}]"
        assert low in seen and high in seen, f"tier {tier}: both ends must be possible"


def test_er_nobody_arrives():
    stats = simulate(num_doctors=3, arrival_prob=0.0, minutes=600, seed=4)
    for tier in TIERS:
        assert stats[tier] == {"arrived": 0, "seen": 0, "total_wait": 0, "still_waiting": 0}


def test_er_everyone_is_counted():
    stats = simulate(num_doctors=4, arrival_prob=0.3, minutes=2000, seed=5)
    assert sum(stats[t]["arrived"] for t in TIERS) > 400, "about 600 arrivals expected"
    for tier in TIERS:
        s = stats[tier]
        assert s["arrived"] == s["seen"] + s["still_waiting"], f"tier {tier}: {s}"


def test_er_plenty_of_doctors_means_no_wait():
    # More doctors than patients: everyone is called in the minute they arrive.
    stats = simulate(num_doctors=1000, arrival_prob=0.5, minutes=500, seed=6)
    for tier in TIERS:
        assert stats[tier]["arrived"] > 0
        assert stats[tier]["total_wait"] == 0, f"tier {tier} waited with 1000 doctors"
        assert stats[tier]["still_waiting"] == 0


def test_er_priority_matters_when_busy():
    # A busy week, but with enough doctors that every tier gets seen.
    stats = simulate(num_doctors=11, arrival_prob=0.25, minutes=SIM_MINUTES, seed=7)
    waits = [average_wait(stats, tier) for tier in TIERS]
    assert None not in waits, f"every tier should have patients seen: {waits}"
    assert waits[0] < 30, f"life-threatening waited {waits[0]:.0f} min on average"
    assert waits[0] < waits[1] < waits[2], f"waits should grow with the tier number: {waits}"


def test_er_same_seed_same_week():
    a = simulate(num_doctors=10, arrival_prob=0.25, minutes=3000, seed=8)
    b = simulate(num_doctors=10, arrival_prob=0.25, minutes=3000, seed=8)
    assert a == b, "the same seed must reproduce the same results"


def test_er_uses_the_heap():
    src = inspect.getsource(simulate)
    assert "MinHeap271()" in src, "the waiting room must be a MinHeap271"
    for banned in ["heapq", "sorted(", ".sort("]:
        assert banned not in src, f"simulate may not use {banned}"


run_tests("test_er")
````
